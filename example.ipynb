# aitool 使い方ガイド

OpenRouter API を使ったマルチモーダル AI 作業を、CLI (`aitool`) から実行する例です。

## 事前準備

1. リポジトリ直下で依存をインストール: `uv sync`
2. `.env.example` をコピーして `.env` を作成し、`OPENROUTER_API_KEY` を設定
3. このノートブックのセルでは、ローカル開発向けに `uv run aitool ...` で実行します

インストール済みの場合は `aitool ...` でも同様に使えます。

## ヘルプの確認

```python
!uv run aitool --help
```

## 画像生成 (generate-image)

テキストのみ指定すると **t2i**（テキスト→画像）、`--image` を付けると **i2i**（画像編集）として動作します。

出力パスは `--output` で明示的に指定します（例: `./generated.png`）。

```python
# t2i: テキストから画像を生成
!uv run aitool generate-image \
  --text "かわいい猫のイラストを生成してください" \
  --output ./generated.png \
  --aspect-ratio 1:1 \
  --image-size 1K
```

```python
# i2i: 既存画像をもとに編集
!uv run aitool generate-image \
  --text "この画像に、かわいい猫を追加してください" \
  --image ./table.png \
  --output ./generated.png
```

## 画像認識 (recognize-image)

`--text` と `--image`（複数可）を指定します。結果は標準出力へ出ます。

```python
!uv run aitool recognize-image \
  --text "この画像を説明してください" \
  --image ./table.png
```

```python
# ファイルへ保存する場合
!uv run aitool recognize-image \
  --text "この画像を説明してください" \
  --image ./table.png \
  --output ./description.txt
```

## 音声文字起こし (transcribe)

既定は STT 専用 API（`--mode dedicated`）です。結果は標準出力へ出ます。

```python
!uv run aitool transcribe --audio ./voice.mp3
```

```python
# ファイルへ保存する場合
!uv run aitool transcribe \
  --audio ./voice.mp3 \
  --output ./transcript.txt
```

```python
# マルチモーダル LLM 経由（プロンプト付き）
!uv run aitool transcribe \
  --mode "llm" \
  --prompt "話者ごとに分けて文字起こししてください" \
  --audio ./meeting.mp3 \
  --model "google/gemini-3-flash-preview"
```

## 音声合成 (tts)

テキストから音声ファイルを生成します。`--output` は必須です。

```python
!uv run aitool tts \
  --text "こんにちは。これは音声合成のテストです。" \
  --output ./tts.mp3 \
  --model "x-ai/grok-voice-tts-1.0" \
  --voice "eve" \
  --format mp3
```

## オプション例

### モデルの一時上書き

`.env` の既定モデルを変えずに、1 回だけ別モデルを使う場合:

```python
!uv run aitool recognize-image \
  --model google/gemini-2.5-pro \
  --text "この画像の内容を箇条書きで" \
  --image ./table.png
```

### API キーを引数で指定

`.env` を使わずに直接渡す場合:

```python
!uv run aitool tts \
  --api-key sk-or-v1-xxxxxxxx \
  --text "テスト" \
  --output ./test.mp3
```

### メタ情報を JSON で表示

`--json` は本文やファイルの代わりに、モデル名や生成 ID などのメタ情報を JSON で出力します。

```python
!uv run aitool generate-image \
  --text "かわいい猫" \
  --output ./generated.png \
  --json
```
